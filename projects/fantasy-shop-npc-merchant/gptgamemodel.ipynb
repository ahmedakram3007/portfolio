```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch
tokenizer = AutoTokenizer.from_pretrained("microsoft/DialoGPT-small")
model = AutoModelForCausalLM.from_pretrained("microsoft/DialoGPT-small")
long_term_memory = (
    "You are an NPC who remembers the player’s name is Melenia, "
    "she completed the forest temple quest, and travels with Luna the wolf."
    "You are a shopkeeper who is selling magical wares and guiding her to a quest to fight an Evil Dragon"
)

long_term_memory_ids = tokenizer.encode(long_term_memory + tokenizer.eos_token, return_tensors='pt')

chat_history_ids = long_term_memory_ids  
for step in range(7):
    user_input = input("You: ")
    input_ids = tokenizer.encode(user_input + tokenizer.eos_token, return_tensors='pt')
    
    bot_input_ids = torch.cat([chat_history_ids, input_ids], dim=-1)
    output_ids = model.generate(bot_input_ids, max_length=1000, pad_token_id=tokenizer.eos_token_id, do_sample=True, top_p=0.9, top_k=50)
    response = tokenizer.decode(output_ids[:, bot_input_ids.shape[-1]:][0], skip_special_tokens=True)
    print("NPC:", response)
    chat_history_ids = torch.cat([bot_input_ids, output_ids[:, bot_input_ids.shape[-1]:]], dim=-1)
```
